```python
import pandas as pd

df = pd.read_csv("Clustered_Customers.csv")

df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
segment_names = {
    0: "Budget Customers",
    1: "Premium Customers",
    2: "Regular Customers",
    3: "VIP Customers"
}
```

```python
df["Customer_Segment"] = df["Cluster"].map(segment_names)
```

```python
df[["Customer_ID", "Cluster", "Customer_Segment"]].head(10)
```

```python
segment_counts = (
    df.groupby(["Cluster", "Customer_Segment"])
      .size()
      .reset_index(name="Customers")
)

segment_counts
```

```python
total_customers = df["Customer_ID"].nunique()

total_spend = df["Total_Spend"].sum()

average_spend = df["Total_Spend"].mean()

average_items = df["Items_Purchased"].mean()

average_rating = df["Average_Rating"].mean()

number_of_segments = df["Cluster"].nunique()
```

```python
print("Total Customers:", total_customers)
print("Total Spend:", round(total_spend, 2))
print("Average Spend:", round(average_spend, 2))
print("Average Items Purchased:", round(average_items, 2))
print("Average Rating:", round(average_rating, 2))
print("Number of Segments:", number_of_segments)
```

# Final Dashboard

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.express as px

# -----------------------------
# Prepare data
# -----------------------------

segment_order = [
    "Budget Customers",
    "Premium Customers",
    "Regular Customers",
    "VIP Customers"
]

segment_counts = (
    df.groupby("Customer_Segment")
      .size()
      .reindex(segment_order)
      .reset_index(name="Customers")
)

total_spending = (
    df.groupby("Customer_Segment")["Total_Spend"]
      .sum()
      .reindex(segment_order)
)

avg_spending = (
    df.groupby("Customer_Segment")["Total_Spend"]
      .mean()
      .reindex(segment_order)
)

avg_items = (
    df.groupby("Customer_Segment")["Items_Purchased"]
      .mean()
      .reindex(segment_order)
)

avg_satisfaction = (
    df.groupby("Customer_Segment")["Satisfaction_Level"]
      .mean()
      .reindex(segment_order)
)

# -----------------------------
# Create dashboard
# -----------------------------

fig = make_subplots(
    rows=4,
    cols=2,

    specs=[
        [
            {"type": "indicator"},
            {"type": "indicator"}
        ],
        [
            {"type": "xy"},
            {"type": "xy"}
        ],
        [
            {"type": "xy"},
            {"type": "xy"}
        ],
        [
            {"type": "xy", "colspan": 2},
            None
        ]
    ],

    subplot_titles=(
        "Total Customers",
        "Total Spend",
        "Customer Distribution",
        "Total Spend by Segment",
        "Average Items Purchased",
        "Average Satisfaction",
        "Customer Spending vs Items Purchased"
    ),

    vertical_spacing=0.10
)

# -----------------------------
# KPI 1
# -----------------------------

fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_customers,
        title={"text": "Customers"}
    ),
    row=1,
    col=1
)

# -----------------------------
# KPI 2
# -----------------------------

fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_spend,
        number={"prefix": "$", "valueformat": ",.2f"},
        title={"text": "Total Spend"}
    ),
    row=1,
    col=2
)

# -----------------------------
# Customer Distribution
# -----------------------------

fig.add_trace(
    go.Bar(
        x=segment_order,
        y=segment_counts["Customers"],
        text=segment_counts["Customers"],
        textposition="auto",
        name="Customers"
    ),
    row=2,
    col=1
)

# -----------------------------
# Total Spend
# -----------------------------

fig.add_trace(
    go.Bar(
        x=segment_order,
        y=total_spending.values,
        text=[f"{x:,.0f}" for x in total_spending.values],
        textposition="auto",
        name="Total Spend"
    ),
    row=2,
    col=2
)

# -----------------------------
# Average Items
# -----------------------------

fig.add_trace(
    go.Bar(
        x=segment_order,
        y=avg_items.values,
        text=[f"{x:.2f}" for x in avg_items.values],
        textposition="auto",
        name="Avg Items"
    ),
    row=3,
    col=1
)

# -----------------------------
# Satisfaction
# -----------------------------

fig.add_trace(
    go.Bar(
        x=segment_order,
        y=avg_satisfaction.values,
        text=[f"{x:.2f}" for x in avg_satisfaction.values],
        textposition="auto",
        name="Satisfaction"
    ),
    row=3,
    col=2
)

# -----------------------------
# Scatter Plot
# -----------------------------

for segment in segment_order:

    temp = df[df["Customer_Segment"] == segment]

    fig.add_trace(
        go.Scatter(
            x=temp["Items_Purchased"],
            y=temp["Total_Spend"],
            mode="markers",
            name=segment,
            text=temp["Customer_ID"],
            hovertemplate=
                "Customer ID: %{text}<br>" +
                "Items: %{x}<br>" +
                "Spend: %{y:.2f}<extra></extra>"
        ),
        row=4,
        col=1
    )

# -----------------------------
# Layout
# -----------------------------

fig.update_layout(
    title={
        "text": "Customer Segmentation Dashboard",
        "x": 0.5
    },
    height=1500,
    showlegend=True
)

fig.update_xaxes(
    title_text="Customer Segment",
    row=2,
    col=1
)

fig.update_xaxes(
    title_text="Customer Segment",
    row=2,
    col=2
)

fig.update_xaxes(
    title_text="Customer Segment",
    row=3,
    col=1
)

fig.update_xaxes(
    title_text="Customer Segment",
    row=3,
    col=2
)

fig.update_xaxes(
    title_text="Items Purchased",
    row=4,
    col=1
)

fig.update_yaxes(
    title_text="Total Spend",
    row=4,
    col=1
)

fig.show()
```
